```python
# Basic Logistic Regression in NumPy
# Applied to Iris dataset

# Zi Wang, 2026

import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split

from LogisticRegression import LogisticRegression



def metrics(df, label_col, prediction_col = 'prediction'):
    '''
    Function to calculate typical classification metrics
    Precision = TP / (TP + FP)
    Recall = TP / (TP + FN)
    F1 score = harmonic mean of precision & recall = 2 * Precision * Recall / (Precision + Recall)
    '''
    df['tp'] = 0
    df['tp'] = df['tp'].case_when([((df[label_col] == 1) & (df[prediction_col] == 1), 1)])

    df['fp'] = 0
    df['fp'] = df['fp'].case_when([((df[label_col] == 1) & (df[prediction_col] == 0), 1)])

    df['tn'] = 0
    df['tn'] = df['tn'].case_when([((df[label_col] == 0) & (df[prediction_col] == 0), 1)])
    
    df['fn'] = 0
    df['fn'] = df['fn'].case_when([((df[label_col] == 0) & (df[prediction_col] == 1), 1)])

    df_agg = df.sum()
    
    # precision
    precision = df_agg['tp'] / (df_agg['tp'] + df_agg['fp'])
    print("Precision:", precision)
    
    # recall
    recall = df_agg['tp'] / (df_agg['tp'] + df_agg['fn'])
    print("Recall:", recall)
    
    # f1
    print("F1 Score:", 2.0 * precision * recall / (precision + recall))

    return df



# Load Iris dataset into pandas dataframe
filename = 'Iris.csv'

df = pd.read_csv(filename)

# set feature and label columns
features_col = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
label_col = 'Species'
prediction_col = 'prediction'

# preprocess dataset to set label for one species to 1 and the others to 0
# simplification to test a binary classifier for one species only
lr_sample_predict = 'Iris-setosa'

df_processed = df[features_col + [label_col]]
df_processed[label_col] = df_processed[label_col].case_when([(df_processed[label_col] == lr_sample_predict, 1), (df_processed[label_col] != lr_sample_predict, 0)])


# Load LR model
lr = LogisticRegression(features_col, label_col)

# Split dataset into train/test samples
df_train, df_test = train_test_split(df_processed, test_size=0.33, random_state=42)

# Train on training set
lr.fit(df_train)

# Predict on test set
predictions = lr.predict(df_test)

# Calculate metrics
metrics(predictions, label_col)


```
